```python
# Calculate Average
def calculate_average(numbers):
    if len(numbers) == 0:
        return 0

    total = sum(numbers)
    average = total / len(numbers)

    return average


numbers = [10, 20, 30, 40, 50]

result = calculate_average(numbers)

print("Average:", result)
```

```python
# find_maximum() without max()
def find_maximum(numbers):
    if len(numbers) == 0:
        return None

    largest = numbers[0]

    for number in numbers:
        if number > largest:
            largest = number

    return largest


numbers = [10, 25, 7, 45, 18]

result = find_maximum(numbers)

print("Maximum:", result)
```

```python
# Modular program — mean and maximum
def read_numbers():
    numbers = input("Enter numbers separated by spaces: ")
    return [float(x) for x in numbers.split()]


def calculate_mean(numbers):
    return sum(numbers) / len(numbers)


def calculate_maximum(numbers):
    largest = numbers[0]

    for number in numbers:
        if number > largest:
            largest = number

    return largest


def display_results(mean, maximum):
    print("Mean:", mean)
    print("Maximum:", maximum)


# Main program
numbers = read_numbers()

mean = calculate_mean(numbers)
maximum = calculate_maximum(numbers)

display_results(mean, maximum)
```

```python
# count_even()
def count_even(numbers):
    count = 0

    for number in numbers:
        if number % 2 == 0:
            count += 1

    return count


numbers = [1, 2, 4, 7, 8, 11, 12]

result = count_even(numbers)

print("Number of even values:", result)
```

```python
# Classification accuracy
def calculate_accuracy(y_true, y_pred):
    correct = 0

    for actual, predicted in zip(y_true, y_pred):
        if actual == predicted:
            correct += 1

    accuracy = correct / len(y_true)

    return accuracy


def display_result(accuracy):
    print("Accuracy:", accuracy)
    print("Accuracy Percentage:", accuracy * 100, "%")


# Example
y_true = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 0]

accuracy = calculate_accuracy(y_true, y_pred)

display_result(accuracy)
```

```python
# Mean and standard deviation
def calculate_mean_std(numbers):
    mean = sum(numbers) / len(numbers)

    squared_difference = 0

    for number in numbers:
        squared_difference += (number - mean) ** 2

    variance = squared_difference / len(numbers)

    standard_deviation = variance ** 0.5

    return mean, standard_deviation


numbers = [10, 20, 30, 40, 50]

mean, std = calculate_mean_std(numbers)

print("Mean:", mean)
print("Standard Deviation:", std)
```

```python
# Min-Max normalization
def normalize(numbers):
    minimum = min(numbers)
    maximum = max(numbers)

    if maximum == minimum:
        return [0 for _ in numbers]

    normalized = []

    for x in numbers:
        value = (x - minimum) / (maximum - minimum)
        normalized.append(value)

    return normalized


numbers = [10, 20, 30, 40, 50]

result = normalize(numbers)

print(result)
```

```python
# Modular ML preprocessing pipeline
def load_data():
    data = [
        [10, 20],
        [15, None],
        [20, 30],
        [25, 40],
        [30, 50]
    ]

    return data


def remove_missing_values(data):
    cleaned_data = []

    for row in data:
        if None not in row:
            cleaned_data.append(row)

    return cleaned_data


def normalize_data(data):
    columns = len(data[0])
    normalized = []

    minimums = []
    maximums = []

    # Find minimum and maximum for each column
    for col in range(columns):
        values = [row[col] for row in data]
        minimums.append(min(values))
        maximums.append(max(values))

    # Normalize each value
    for row in data:
        new_row = []

        for col in range(columns):
            minimum = minimums[col]
            maximum = maximums[col]

            if maximum == minimum:
                value = 0
            else:
                value = (row[col] - minimum) / (maximum - minimum)

            new_row.append(value)

        normalized.append(new_row)

    return normalized


def display_data(data):
    for row in data:
        print(row)


# Pipeline
data = load_data()

data = remove_missing_values(data)

data = normalize_data(data)

display_data(data)
```

```python
# Euclidean distance
def euclidean_distance(point1, point2):
    x1, y1 = point1
    x2, y2 = point2

    distance = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5

    return distance


point1 = (2, 3)
point2 = (5, 7)

result = euclidean_distance(point1, point2)

print("Euclidean Distance:", result)
```
